# Generate mock subjects for PSD data

```python
from pathlib import Path
import pandas as pd
```

```python
data_folder = "."
data_csv_file = "psd_summary.csv"

output_folder = "../.."
output_csv_file = "dataset_PSD.csv"
```

```python
df_original = pd.read_csv(Path(data_folder) / data_csv_file)
```

```python
subj_id = [f"sub-{id+1:03}" for id in range(len(df_original))]
df_original.subject_id = subj_id
```

```python
df_original.to_csv(Path(output_folder) / output_csv_file, index=False)
```
